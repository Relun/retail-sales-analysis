## Columnas:

**Transaction ID**: un identificador único para cada transacción, que permite su seguimiento y consulta.

**Date**: la fecha en la que se realizó la transacción, lo que ofrece información sobre las tendencias de ventas a lo largo del tiempo.

**Customer ID**: un identificador único para cada cliente, que permite realizar análisis centrados en el cliente.

**Gender**: El género del cliente (hombre/mujer), lo que ofrece información sobre los patrones de compra según el género.

**Age**: La edad del cliente, lo que facilita la segmentación y el análisis de las influencias relacionadas con la edad.

**Product Category**: La categoría del producto comprado (por ejemplo, electrónica, ropa, belleza), lo que ayuda a comprender las preferencias de producto.

**Quantity**: El número de unidades del producto comprado, lo que contribuye a obtener información sobre los volúmenes de compra.


**Price per Unit**: El precio de una unidad del producto, que ayuda en los cálculos relacionados con el gasto total.

**Total Amount**: El valor monetario total de la transacción, que muestra el impacto financiero de cada compra.

## Transformación de Datos

```python
df2 = df.copy()
```

Primero pasamos la fecha a tipo datetime, ya que actualmente está en object.

```python
df2['Date'] = pd.to_datetime(df2['Date'])
```

```python
df2.dtypes
```

**Crea nuevas columnas**: **Basándonos en los datos existentes, crea nuevas columnas que sean útiles para el análisis. Por ejemplo, calcula el ingreso total por venta y normaliza las ventas.**

Le sumaremos el IVA  las ventas

```python
df2['Total Amount IVA'] = df2['Total Amount'] * 1.19
df2
```

Normalización de Total Amount

```python
df2['Total Amount Normalizado'] = (df2['Total Amount'] - df2['Total Amount'].min() / (df2['Total Amount'].max() - df2['Total Amount'].min())).round(1)
df2
```

**Clasifica los datos**: **Crea una columna que clasifique las ventas en categorías significativas (e.g., ‘Alta’, ‘Media’, ‘Baja’).**

Primero vemos cual es el valor máximo y mínimo de ventas para hacer la clasificación

```python
df2['Total Amount'].max()
```

```python
df2['Total Amount'].min()
```

```python
def categoria_ventas(total_amount):
  if total_amount < 700:
    return 'Baja'
  elif total_amount < 1000:
    return 'Media'
  else:
    return 'Alta'
```

```python
df2['Total Amount Categoría'] = df2['Total Amount'].apply(categoria_ventas)
df2
```

```python
reorden_columnas =['Transaction ID', 'Date', 'Customer ID', 'Gender', 'Age', 'Product Category', 'Quantity', 'Price per Unit', 'Total Amount', 'Total Amount Categoría', 'Total Amount IVA', 'Total Amount Normalizado' ]
```

```python
df2 = df2[reorden_columnas]
df2
```

## Agrupación y Agregación

**Agrupación por múltiples columnas**: Realiza agrupaciones por categorías como producto y tienda, producto y mes, etc.

Sumatoria de ventas por cada categoría

```python
producto_y_ventas = df2.groupby('Product Category')['Total Amount'].sum()
producto_y_ventas
```

Ventas totales por genero.

```python
genero_y_ventas = df2.groupby('Gender')['Total Amount'].sum()
genero_y_ventas
```

**Aplicar funciones de agregación**: Utiliza funciones como sum, mean, count, min, max, std, y var para obtener estadísticas descriptivas de cada grupo.

```python
df2.groupby('Total Amount Categoría')['Total Amount'].agg(['mean', 'count', 'min', 'max', 'std', 'var'])
```

## Análisis Personalizado con apply

**Función personalizada**: Aplica funciones personalizadas para realizar análisis específicos que no se pueden lograr con las funciones de agregación estándar.

Vemos las edades mínimas y máximas para hacer la clasificación por edad

```python
df2['Age'].max()
```

```python
df2['Age'].min()
```

```python
def categoria_edad(age):
  if age < 18:
    return "Menor de edad"
  elif age < 30:
    return "Joven adulto"
  elif age < 60:
    return "Adulto"
  else:
    return "Adulto mayor"
```

```python
df2['Age Category'] = df2['Age'].apply(categoria_edad)
df2
```

```python
reordenar_columnas =['Transaction ID', 'Date', 'Customer ID', 'Gender', 'Age', 'Age Category', 'Product Category', 'Quantity', 'Price per Unit', 'Total Amount', 'Total Amount Categoría', 'Total Amount IVA', 'Total Amount Normalizado' ]
```

```python
df2 = df2[reordenar_columnas]
df2
```

**Ejemplo de uso avanzado**: Calcula la desviación de cada venta respecto a la media de su grupo.

**Nota: aquí le pedí ayuda a la IA porque no entendía que me preguntaban.

Compararemos cada venta con el promedio del grupo al que pertenece (valor individual - promedio del grupo)

```python
df2['Desviacion'] = ((df2['Total Amount'] - df2.groupby('Total Amount Categoría')['Total Amount'].transform('mean'))).round(2)
```

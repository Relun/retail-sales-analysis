## Columnas:

**Transaction ID**: un identificador único para cada transacción, que permite su seguimiento y consulta.

**Date**: la fecha en la que se realizó la transacción, lo que ofrece información sobre las tendencias de ventas a lo largo del tiempo.

**Customer ID**: un identificador único para cada cliente, que permite realizar análisis centrados en el cliente.

**Gender**: El género del cliente (hombre/mujer), lo que ofrece información sobre los patrones de compra según el género.

**Age**: La edad del cliente, lo que facilita la segmentación y el análisis de las influencias relacionadas con la edad.

**Product Category**: La categoría del producto comprado (por ejemplo, electrónica, ropa, belleza), lo que ayuda a comprender las preferencias de producto.

**Quantity**: El número de unidades del producto comprado, lo que contribuye a obtener información sobre los volúmenes de compra.


**Price per Unit**: El precio de una unidad del producto, que ayuda en los cálculos relacionados con el gasto total.

**Total Amount**: El valor monetario total de la transacción, que muestra el impacto financiero de cada compra.

**Exploración de Datos Avanzada**


Realiza un análisis de correlación y crea un mapa de calor para visualizar las relaciones entre las variables.


Crea subplots para comparar diferentes variables clave.


Añade cuadrículas, leyendas, anotaciones y flechas a tus gráficos para mejorar la claridad y la información.

```python
cols_num = ["Age", "Quantity", "Price per Unit", "Total Amount"]
corr = df2[cols_num].corr()

fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(
    corr, annot=True, fmt=".2f",
    cmap="coolwarm", center=0, vmin=-1, vmax=1,
    square=True, linewidths=0.5,
    annot_kws={"size": 11}, ax=ax,
    cbar_kws={"shrink": 0.8}
)
ax.set_title("Matriz de Correlación — Ventas", fontsize=13, fontweight="bold", pad=16)
plt.tight_layout()
plt.show()
```

La correlación positiva más fuerte del gráfico se da entre Price per Unit y Total Amount (0.85), esto quiere decir que cuando aumente el precio por unidad también aumentará el monto total de compra.

Luego le sigue la correlación Quantity y Total Amount (0.37), esto significa que a mayor cantidad comprada, el monto total tiende a aumentar, sin embargo, la cantidad no influye tanto en el monto total, como lo hace el precio por unidad.

Las demás relaciones son practicamente cero, por lo tanto podemos deducir que no hay relación entre ellas.

```python
fig, ax = plt.subplots(1, 2, figsize=(12,5))
#histograma
ax[0].hist(df2["Total Amount"], bins='auto', color="gold", edgecolor="white", alpha=0.85, label="Frecuencia")

media   = df2["Total Amount"].mean()
mediana = df2["Total Amount"].median()
ax[0].axvline(media,   color="#00C9A7", linewidth=2.5, linestyle="--", label=f"Media: {media:.1f}")
ax[0].axvline(mediana, color="#F59E0B", linewidth=2.5, linestyle=":",  label=f"Mediana: {mediana:.1f}")

ax[0].annotate("Promedio", xy=(media, 300), xytext=(media, 350), arrowprops=dict(arrowstyle="->", color="red"))

ax[0].set_title("Distribución de las ventas")
ax[0].set_xlabel("Ventas", fontsize=12)
ax[0].set_ylabel("Frecuencia", fontsize=12)
ax[0].legend(fontsize=9, bbox_to_anchor=(1, 1), loc="upper left")
ax[0].grid(axis="y", alpha=0.25)

#boxplot
ax[1].boxplot(df2["Total Amount"], patch_artist=True, boxprops=dict(facecolor="royalblue"))
ax[1].set_title("Boxplot de ventas")
ax[1].set_ylabel("Ventas")
ax[1].set_xticks([])

plt.tight_layout()
plt.show()
```

El histograma muestra que la mayoría de las ventas son por montos pequeños. La media está muy por sobre la mediana debido a ventas de elevado valor.

El 50% de las ventas son menores a 135 (mediana), el bigote de arriba se extiende hasta los 2000 demostrando la variabilidad de los datos, a pesar de ello, no existen valores atípicos.

```python
fig, ax = plt.subplots(1, 2, figsize=(12,5))
#histograma
ax[0].hist(df2["Age"], bins='auto', color="powderblue", edgecolor="white", alpha=0.85, label="Frecuencia")

media   = df2["Age"].mean()
mediana = df2["Age"].median()
ax[0].axvline(media,   color="#00C9A7", linewidth=2.5, linestyle="--", label=f"Media: {media:.1f}")
ax[0].axvline(mediana, color="#F59E0B", linewidth=2.5, linestyle=":",  label=f"Mediana: {mediana:.1f}")
ax[0].annotate("Promedio", xy=(media, 100), xytext=(media, 120), arrowprops=dict(arrowstyle="->", color="red"))

ax[0].set_title("Distribución de la edad")
ax[0].set_xlabel("Edad", fontsize=12)
ax[0].set_ylabel("Frecuencia", fontsize=12)
ax[0].legend(fontsize=9, bbox_to_anchor=(1, 1), loc="upper left")
ax[0].grid(axis="y", alpha=0.25)

#boxplot
ax[1].boxplot(df2["Age"], patch_artist=True, boxprops=dict(facecolor="silver"))
ax[1].set_title("Boxplot de edad")
ax[1].set_ylabel("Edad")
ax[1].set_xticks([])

plt.tight_layout()
plt.show()
```

Las edades de los clientes muestran una distribución equilibrada, con una leve tendencia entre los más jóvenes, en la parte media y entre los más mayores. La media es levemente más baja que la mediana, lo cuál confirma esta distribución equilibrada. El boxplot también muestra una distribución equilibrada, sin outliers.

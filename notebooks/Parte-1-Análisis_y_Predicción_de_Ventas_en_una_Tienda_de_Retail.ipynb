**Transaction ID**: un identificador único para cada transacción, que permite su seguimiento y consulta.

**Date**: la fecha en la que se realizó la transacción, lo que ofrece información sobre las tendencias de ventas a lo largo del tiempo.

**Customer ID**: un identificador único para cada cliente, que permite realizar análisis centrados en el cliente.

**Gender**: El género del cliente (hombre/mujer), lo que ofrece información sobre los patrones de compra según el género.

**Age**: La edad del cliente, lo que facilita la segmentación y el análisis de las influencias relacionadas con la edad.

**Product Category**: La categoría del producto comprado (por ejemplo, electrónica, ropa, belleza), lo que ayuda a comprender las preferencias de producto.

**Quantity**: El número de unidades del producto comprado, lo que contribuye a obtener información sobre los volúmenes de compra.


**Price per Unit**: El precio de una unidad del producto, que ayuda en los cálculos relacionados con el gasto total.

**Total Amount**: El valor monetario total de la transacción, que muestra el impacto financiero de cada compra.

Calcula el total de ventas por categoría de producto.


```python
total_de_ventas_por_categoria = df.groupby('Product Category')['Total Amount'].sum()
total_de_ventas_por_categoria
```

La venta de electronics recaudó más dinero que la venta de los otros productos

Tarea: Calcula el promedio de ventas diarias por categoría de producto.


```python
promedio_ventas_diarias_por_categoria = df.groupby(['Product Category', 'Date'])['Total Amount'].mean()
promedio_ventas_diarias_por_categoria
```

Tarea: Identifica las categorías de productos con mayores y menores ventas.

*Nota: Aquí pedí ayuda a la IA porque no entendía que me estaban pidiendo.

```python
producto_con_mayor_venta = total_de_ventas_por_categoria.max()
categoria_mayor_venta = total_de_ventas_por_categoria.idxmax()

print(f"Categoria con mayor venta: {categoria_mayor_venta} y su total: {producto_con_mayor_venta}")
```

```python
producto_con_menor_venta = total_de_ventas_por_categoria.min()
categoria_menor_venta = total_de_ventas_por_categoria.idxmin()

print(f"Categoria con menor venta: {categoria_menor_venta} y su total: {producto_con_menor_venta}")
```

```python
cantidades_por_categoria = df.groupby('Product Category')['Quantity'].sum() #detalle de las cantidades vendidas por cada categoria
cantidades_por_categoria
```

La ropa fue la categoría más vendida por unidad

```python
transacciones_por_genero = df['Gender'].value_counts()
transacciones_por_genero
```

Las mujeres hicieron hicieron más compras que los hombres

```python
compras_por_genero = df.groupby('Gender')['Total Amount'].sum()
compras_por_genero
```

Las mujeres gastaron más que los hombres

```python
gasto_promedio_por_compra = df.groupby('Gender')['Total Amount'].mean()
gasto_promedio_por_compra
```

En promedio las mujeres gastaron un poco más que los hombres

```python
total_ventas = df['Total Amount'].sum()
print(f"La cantidad total de todas las ventas es: {total_ventas}")
```

```python
cantidad_electronics = cantidades_por_categoria['Electronics']
print(f"La cantidad de productos electrónicos vendidos es: {cantidad_electronics}")
```

```python
suma = df['Quantity'].sum()
print(f"La cantidad total de productos vendidos es: {suma}")

resta = suma - cantidad_electronics
print(f"Las cantidades vendidas entre Beauty y Clothing es: {resta}")

multiplicacion = total_ventas * 0.9
print(f"El total de ventas con un 10% de descuento es: {multiplicacion}")
```

## Columnas:

**Transaction ID**: un identificador único para cada transacción, que permite su seguimiento y consulta.

**Date**: la fecha en la que se realizó la transacción, lo que ofrece información sobre las tendencias de ventas a lo largo del tiempo.

**Customer ID**: un identificador único para cada cliente, que permite realizar análisis centrados en el cliente.

**Gender**: El género del cliente (hombre/mujer), lo que ofrece información sobre los patrones de compra según el género.

**Age**: La edad del cliente, lo que facilita la segmentación y el análisis de las influencias relacionadas con la edad.

**Product Category**: La categoría del producto comprado (por ejemplo, electrónica, ropa, belleza), lo que ayuda a comprender las preferencias de producto.

**Quantity**: El número de unidades del producto comprado, lo que contribuye a obtener información sobre los volúmenes de compra.


**Price per Unit**: El precio de una unidad del producto, que ayuda en los cálculos relacionados con el gasto total.

**Total Amount**: El valor monetario total de la transacción, que muestra el impacto financiero de cada compra.

Crea gráficos de líneas para mostrar las tendencias de ventas a lo largo del tiempo.


```python
fecha_agrupada_por_mes = df2.groupby(df2['Date'].dt.to_period('M'))['Total Amount'].sum()
fecha_agrupada_por_mes
```

**Nota: Aquí le pedí ayuda a la IA para acomodar el gráfico.

```python
fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(fecha_agrupada_por_mes.index.to_timestamp(), fecha_agrupada_por_mes.values, color="#6C3CE1", linewidth=2.5,
        marker="o", markersize=8, markerfacecolor="#00C9A7")

ax.set_title("Tendencias de ventas a lo largo del tiempo", fontsize=14, fontweight="bold", pad=14)
ax.set_xlabel("Fecha", fontsize=12)
ax.set_ylabel("Monto total de ventas", fontsize=12)
ax.grid(alpha=0.3, linestyle="--")
plt.xticks(rotation=45) # Rotate x-axis labels
plt.tight_layout()
plt.show()
```

El gráfico muestra las ventas mensuales durante todo un año, las cuáles son variables, sin existir una tendencia lineal, vemos que el mayor nivel de ventas se dio en 05-2023, por otro lado hay una drástica caída en las ventas en 01-2024.

Combina histogramas y boxplots para visualizar la distribución de las ventas y detectar valores atípicos.

```python
fig, ax = plt.subplots(1, 2, figsize=(12,5))
#histograma
ax[0].hist(df["Total Amount"], bins='auto', color="gold", edgecolor="white", alpha=0.85, label="Frecuencia")

media   = df["Total Amount"].mean()
mediana = df["Total Amount"].median()
ax[0].axvline(media,   color="#00C9A7", linewidth=2.5, linestyle="--", label=f"Media: {media:.1f}")
ax[0].axvline(mediana, color="#F59E0B", linewidth=2.5, linestyle=":",  label=f"Mediana: {mediana:.1f}")

ax[0].set_title("Distribución de las ventas")
ax[0].set_xlabel("Ventas", fontsize=12)
ax[0].set_ylabel("Frecuencia", fontsize=12)
ax[0].legend(fontsize=9, bbox_to_anchor=(1, 1), loc="upper left")
ax[0].grid(axis="y", alpha=0.25)

#boxplot
ax[1].boxplot(df["Total Amount"], patch_artist=True, boxprops=dict(facecolor="royalblue"))
ax[1].set_title("Boxplot de ventas")
ax[1].set_ylabel("Ventas")
ax[1].set_xticks([])

plt.tight_layout()
plt.show()
```

El histograma muestra que la mayoría de las ventas son por montos pequeños. La media está muy por sobre la mediana debido a ventas de elevado valor.

El 50% de las ventas son menores a 135 (mediana), el bigote de arriba se extiende hasta los 2000 demostrando la variabilidad de los datos, a pesar de ello, no existen valores atípicos.

Utiliza gráficos de dispersión para analizar la relación entre diferentes variables.


**Nota, en estos gráficos de dispersión pedí ayuda a la IA para entender cómo interpretarlos.

```python
fig, ax = plt.subplots(figsize=(9, 6))

ax.scatter(df2['Age'], df2['Total Amount'], alpha=0.35, color="#6C3CE1", s=25)

ax.set_title("Cómo afecta la edad de los clientes en el monto total de las ventas", fontsize=14, fontweight="bold")

ax.set_xlabel("Edad", fontsize=12)
ax.set_ylabel("Monto total de ventas", fontsize=12)
ax.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

Según el gráfico, no existe relación entre la edad y las ventas.

```python
fig, ax = plt.subplots(figsize=(9, 6))

ax.scatter(df2['Price per Unit'], df2['Total Amount'], alpha=0.35, color="maroon", s=25)

ax.set_title("Cómo afecta el precio por unidad en las ventas", fontsize=14, fontweight="bold")

ax.set_xlabel("Precio por unidad", fontsize=12)
ax.set_ylabel("Monto total de ventas", fontsize=12)
ax.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

Aquí vemos una relación, a mayor precio por unidad, mayor es el monto total de las ventas.

Utiliza histogramas y boxplots para visualizar la distribución de las variables numéricas clave.


```python
fig, ax = plt.subplots(1, 2, figsize=(12,5))
#histograma
ax[0].hist(df["Age"], bins='auto', color="powderblue", edgecolor="white", alpha=0.85, label="Frecuencia")

media   = df["Age"].mean()
mediana = df["Age"].median()
ax[0].axvline(media,   color="#00C9A7", linewidth=2.5, linestyle="--", label=f"Media: {media:.1f}")
ax[0].axvline(mediana, color="#F59E0B", linewidth=2.5, linestyle=":",  label=f"Mediana: {mediana:.1f}")

ax[0].set_title("Distribución de la edad")
ax[0].set_xlabel("Edad", fontsize=12)
ax[0].set_ylabel("Frecuencia", fontsize=12)
ax[0].legend(fontsize=9, bbox_to_anchor=(1, 1), loc="upper left")
ax[0].grid(axis="y", alpha=0.25)

#boxplot
ax[1].boxplot(df["Age"], patch_artist=True, boxprops=dict(facecolor="silver"))
ax[1].set_title("Boxplot de edad")
ax[1].set_ylabel("Edad")
ax[1].set_xticks([])

plt.tight_layout()
plt.show()
```

Las edades de los clientes muestran una distribución equilibrada, con una leve tendencia entre los más jóvenes, en la parte media y entre los más mayores. La media es levemente más baja que la mediana, lo cuál confirma esta distribución equilibrada. El boxplot también muestra una distribución equilibrada, sin outliers.

```python
fig, ax = plt.subplots(1, 2, figsize=(12,5))
#histograma
ax[0].hist(df["Price per Unit"], bins='auto', color="palegreen", edgecolor="white", alpha=0.85, label="Frecuencia")

media   = df["Price per Unit"].mean()
mediana = df["Price per Unit"].median()
ax[0].axvline(media,   color="#00C9A7", linewidth=2.5, linestyle="--", label=f"Media: {media:.1f}")
ax[0].axvline(mediana, color="#F59E0B", linewidth=2.5, linestyle=":",  label=f"Mediana: {mediana:.1f}")

ax[0].set_title("Distribución del precio por unidad")
ax[0].set_xlabel("Precio por unidad", fontsize=12)
ax[0].set_ylabel("Frecuencia", fontsize=12)
ax[0].legend(fontsize=9, bbox_to_anchor=(1, 1), loc="upper left")
ax[0].grid(axis="y", alpha=0.25)

#boxplot
ax[1].boxplot(df["Price per Unit"], patch_artist=True, boxprops=dict(facecolor="pink"))
ax[1].set_title("Boxplot del precio por unidad")
ax[1].set_ylabel("Precio por unidad")
ax[1].set_xticks([])

plt.tight_layout()
plt.show()
```

Las barras separadas del histograma se debe a que el precio por unidad sólo presenta 5 valores (25, 30, 50, 300, 500). La media está muy por sobre la mediana, a pesar de que los precios por unidad en su mayoría están en valores bajos, los precios elevados hacen que la media suba. El boxplot confirma la variabilidad del rango de precios por unidad, sin outliers.
